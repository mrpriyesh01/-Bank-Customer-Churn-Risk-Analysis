```python
#Age Group


df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 30, 40, 50, 60, 100],
    labels=["Under 30", "30-40", "40-50", "50-60", "60+"],
    right=False
)
```

```python
#Create Risk Level
df["RiskLevel"] = pd.cut(
    df["RiskScore"],
    bins=[-1, 1, 3, 6],
    labels=["Low Risk", "Medium Risk", "High Risk"]
)
```

```python
#Create RiskScore

df["RiskScore"] = (
    (df["IsActiveMember"] == 0).astype(int)
    + (df["Age"] >= 40).astype(int)
    + (df["CreditScore"] < 600).astype(int)
    + (df["NumOfProducts"] == 1).astype(int)
    + (df["NumOfProducts"] >= 3).astype(int)
    + (df["Balance"] == 0).astype(int)
)
```

```python
df[["IsActiveMember", "Age", "CreditScore", "NumOfProducts", "Balance", "RiskScore"]].head(10)
```

```python
#Now create RiskLevel
df["RiskLevel"] = pd.cut(
    df["RiskScore"],
    bins=[-1, 1, 3, 6],
    labels=["Low Risk", "Medium Risk", "High Risk"]
)
```

```python
df[["RiskScore", "RiskLevel"]].head(20)
```

```python
df["RiskLevel"].value_counts()
```

```python
#Check whether the risk score makes sense
df.groupby("RiskLevel", observed=True)["Exited"].mean() * 100
```

```python
#Create a simple customer-risk summary
risk_summary = df.groupby(
    "RiskLevel",
    observed=True
).agg(
    Customers=("CustomerId", "count"),
    Churned_Customers=("Exited", "sum"),
    Avg_Balance=("Balance", "mean")
).reset_index()
```

```python
risk_summary["Churn_Rate"] = (
    risk_summary["Churned_Customers"]
    / risk_summary["Customers"]
) * 100
```

```python
#Credit Score Band
df["CreditBand"] = pd.cut(
    df["CreditScore"],
    bins=[0, 600, 700, 800, 1000],
    labels=["Low", "Fair", "Good", "Excellent"],
    right=False
)
```
